```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("data.csv")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
print("First 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("Dataset information:")
df.info()

print("\nMissing values per column:")
missing_values = df.isnull().sum()

display(
    missing_values.to_frame("Missing Values")
)

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
pd.set_option("display.max_columns", None)

numerical_summary = df.describe().T

display(numerical_summary)
```

```python
# CELL 5 — CATEGORICAL ANALYSIS


categorical_columns = df.select_dtypes(
    include=["object", "bool"]
).columns

print("Categorical / Boolean columns:")
print(categorical_columns.tolist())

print("\nCategorical summary:")
display(
    df[categorical_columns].describe().T
)
```

```python
for column in categorical_columns:

    print("\n" + "=" * 60)
    print(column)

    print("Number of unique values:",
          df[column].nunique())

    print("Unique values:")

    if df[column].nunique() <= 20:
        print(df[column].unique())
    else:
        print(
            df[column].unique()[:20],
            "... "
        )
```

```python
# CELL 6 — DUPLICATE ANALYSIS
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)

if duplicate_count > 0:
    print("\nDuplicate rows:")
    display(df[df.duplicated(keep=False)].sort_values(
        by=df.columns.tolist()
    ))
else:
    print("No exact duplicate rows found.")
```

```python
# CELL 7 — CARDINALITY ANALYSIS

cardinality = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [
        df[column].nunique()
        for column in df.columns
    ],
    "Total Rows": len(df)
})

cardinality["Unique %"] = (
    cardinality["Unique Values"]
    / len(df)
    * 100
)

display(cardinality)
```

```python
# CELL 8 — INVALID VALUE CHECKS
print("INVALID VALUE CHECKS")
print("=" * 60)

# Age
print(
    "Age < 18:",
    (df["customer_age"] < 18).sum()
)

print(
    "Age > 100:",
    (df["customer_age"] > 100).sum()
)


# Tenure
print(
    "Negative tenure:",
    (df["customer_tenure_years"] < 0).sum()
)


# Payment rate
print(
    "Payment rate < 0:",
    (df["on_time_payment_rate"] < 0).sum()
)

print(
    "Payment rate > 100:",
    (df["on_time_payment_rate"] > 100).sum()
)


# Counts
count_columns = [
    "late_payment_count",
    "num_claims_last_year",
    "rejected_claims",
    "claims_approved"
]

for column in count_columns:

    print(
        f"Negative {column}:",
        (df[column] < 0).sum()
    )


# Monetary values
money_columns = [
    "premium_amount",
    "previous_premium_amount",
    "total_claim_amount_last_year"
]

for column in money_columns:

    print(
        f"Negative {column}:",
        (df[column] < 0).sum()
    )
```

```python
# ============================================
# CELL 9 — LOGICAL CONSISTENCY
# ============================================

print("LOGICAL CONSISTENCY CHECKS")
print("=" * 60)


# ------------------------------------------------
# 1. Late payment flag
# ------------------------------------------------

late_payment_check = (
    df["has_late_payments"]
    ==
    (df["late_payment_count"] > 0)
)

print(
    "Late payment flag consistent:",
    late_payment_check.all()
)

print(
    "Inconsistent records:",
    (~late_payment_check).sum()
)


# ------------------------------------------------
# 2. Premium increase %
# ------------------------------------------------

calculated_increase = (
    (
        df["premium_amount"]
        -
        df["previous_premium_amount"]
    )
    /
    df["previous_premium_amount"]
    * 100
)

premium_check = np.isclose(
    df["premium_increase_pct"],
    calculated_increase,
    atol=0.02
)

print(
    "Premium increase % consistent:",
    premium_check.all()
)

print(
    "Inconsistent records:",
    (~premium_check).sum()
)
```

```python
# ============================================
# CELL 10 — ID / REDUNDANCY INVESTIGATION
# ============================================

print("IDENTIFIER ANALYSIS")
print("=" * 60)


# ------------------------------------------------
# Customer ID
# ------------------------------------------------

print(
    "Customer IDs:",
    df["customer_id"].nunique()
)

print(
    "Total records:",
    len(df)
)

print(
    "Customer ID unique:",
    df["customer_id"].nunique() == len(df)
)


# ------------------------------------------------
# Policy ID
# ------------------------------------------------

print(
    "\nPolicy IDs:",
    df["policy_id"].nunique()
)

print(
    "Policy names:",
    df["policy_name"].nunique()
)


# ------------------------------------------------
# Policy ID → Policy Name
# ------------------------------------------------

policy_mapping = (
    df.groupby("policy_id")["policy_name"]
    .nunique()
)

print(
    "\nPolicy IDs mapped to multiple names:",
    (policy_mapping > 1).sum()
)


# ------------------------------------------------
# Policy Name → Policy Type
# ------------------------------------------------

policy_type_mapping = (
    df.groupby("policy_name")["policy_type"]
    .nunique()
)

print(
    "Policy names mapped to multiple types:",
    (policy_type_mapping > 1).sum()
)


# ------------------------------------------------
# Direct equality check
# ------------------------------------------------

print(
    "\npolicy_name == policy_type:",
    (
        df["policy_name"]
        ==
        df["policy_type"]
    ).all()
)
```

```python
from sklearn.utils import resample

# Separate each offer class
df_installment = df[df["offer_given"] == "Installment Payment"]
df_loyalty = df[df["offer_given"] == "Loyalty Discount"]
df_no_claim = df[df["offer_given"] == "No-Claim Protection Benefit"]

# Target size = size of minority class
target_size = len(df_installment)

# Undersample the larger classes
df_loyalty_balanced = resample(
    df_loyalty,
    replace=False,
    n_samples=target_size,
    random_state=42
)

df_no_claim_balanced = resample(
    df_no_claim,
    replace=False,
    n_samples=target_size,
    random_state=42
)

# Keep all minority-class records
df_installment_balanced = df_installment.copy()

# Combine
df_balanced = pd.concat([
    df_installment_balanced,
    df_loyalty_balanced,
    df_no_claim_balanced
])

# Shuffle
df_balanced = df_balanced.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("Original shape:", df.shape)
print("Balanced shape:", df_balanced.shape)

print("\nBalanced class distribution:")
print(df_balanced["offer_given"].value_counts())
```

```python
import matplotlib.pyplot as plt

counts = df_balanced["offer_given"].value_counts()

plt.figure(figsize=(9, 5))

bars = plt.bar(
    counts.index,
    counts.values
)

plt.title("Balanced Distribution of Offer Types")
plt.xlabel("Offer Given")
plt.ylabel("Count")
plt.xticks(rotation=20)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
from imblearn.over_sampling import SMOTENC
from sklearn.model_selection import train_test_split

# Separate features and target
X = df.drop(columns=["offer_given"])
y = df["offer_given"]

# Split BEFORE sampling
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Categorical columns
categorical_features = [
    "customer_gender",
    "customer_occupation",
    "policy_name",
    "payment_frequency"
]

# Get their column indices
categorical_indices = [
    X_train.columns.get_loc(col)
    for col in categorical_features
]

# SMOTENC
smote_nc = SMOTENC(
    categorical_features=categorical_indices,
    random_state=42
)

# Oversample ONLY training data
# Drop identifier columns and target leakage before SMOTENC.
# These columns are raw strings / target-derived values and will break model training.
drop_cols = ["policy_id", "customer_id", "policy_type", "offer_encoded"]

X_train = X_train.drop(columns=drop_cols)
X_test = X_test.drop(columns=drop_cols)

# Oversample ONLY the training data
# Drop non-feature columns first, then recompute categorical positions
# using the current X_train columns; otherwise the SMOTENC indexes are stale.
X_train = X_train.drop(columns=drop_cols, errors="ignore")
X_test = X_test.drop(columns=drop_cols, errors="ignore")

categorical_cols = [
    "customer_gender",
    "customer_occupation",
    "policy_name",
    "payment_frequency"
]

categorical_features = [
    X_train.columns.get_loc(col)
    for col in categorical_cols
    if col in X_train.columns
]

smote_nc = SMOTENC(
    categorical_features=categorical_features,
    random_state=42
)

X_train_resampled, y_train_resampled = smote_nc.fit_resample(
    X_train,
    y_train
)

# Rebuild a dataframe for plotting / downstream checks
df_oversampled = X_train_resampled.copy()
df_oversampled["offer_given"] = y_train_resampled

# Check distributions
print("Original training distribution:")
print(y_train.value_counts().sort_index())

print("\nAfter SMOTENC:")
print(y_train_resampled.value_counts().sort_index())

print("\nOriginal training rows:", len(X_train))
print("After SMOTENC rows:", len(X_train_resampled))

print("\nTest distribution:")
print(y_test.value_counts().sort_index())

# Optional visualization
plt.figure(figsize=(9, 5))
counts = df_oversampled["offer_given"].value_counts()
bars = plt.bar(counts.index, counts.values, color=["#66c2a5", "#fc8d62", "#8da0cb"])

plt.title("Oversampled Distribution of Offer Types")
plt.xlabel("Offer Given")
plt.ylabel("Count")
plt.xticks(rotation=20)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```
